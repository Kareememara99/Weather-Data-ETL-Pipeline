```python
list_cities = [] 

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "appid": "6a13e61e853763292101962a026aa3e8",
    "q": "city"
}

for city in cities:
    params["q"] = city
    response = requests.get(url, params=params)
    data = response.json()
    
    city = city
    country = 'Egypt'
    latitude  =  data['coord']['lat']
    longitude  =  data['coord']['lon']
    tempreature = data['main']['temp'] - 273
    pressure = data['main']['pressure']
    Humidity = data['main']['humidity']
    wind_speed = data['wind']['speed']
    WeatherMain = data['weather'][0]['main']
    weather_description = data['weather'][0]['description']

    list_cities.append([city, country, latitude, longitude, tempreature, pressure, Humidity, wind_speed, WeatherMain, weather_description])
```

```python
list_cities
```

```python
print(type(list_cities))
print(type(list_cities[0]))
print(list_cities[0])
```

```python
import pandas as pd
df = pd.DataFrame(list_cities, columns=['City', 'Country', 'Latitude', 'Longitude', 'Temperature', 'Pressure', 'Humidity', 'Wind Speed', 'Weather Main', 'Weather Description'])
```

```python
df
```

```python
def fun1(city):
    if city in ['Luxor','Aswan']:
        return 'Lower'
    else:
        return 'Upper'
    
```

```python
df['Region'] = df['City'].apply(fun1)
```

```python
df
```

```python
def tempCategory(temp):
    if temp < 20:
        return 'Cold'
    elif temp >= 20 and temp < 30:
        return 'Moderate'
    else:
        return 'Hot'
```

```python
df['Temperature Category'] = df['Temperature'].apply(tempCategory)
```

```python
def HumidityCategory(humidity):
    if humidity < 30:
        return 'Low'
    elif humidity >= 30 and humidity < 60:
        return 'Moderate'
    else:
        return 'High'
```

```python
df['Humidity Category'] = df['Humidity'].apply(HumidityCategory)
```

```python
df
```

```python
import datetime
date = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
print(date)
```

```python
df['Ingestion Date'] = date
```
